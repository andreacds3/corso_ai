```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider
from IPython.display import display, clear_output

# 1. Generazione dati
np.random.seed(42)
area = np.random.uniform(50, 250, 100)
prezzo_reale = (area * 2000) + 50000 + np.random.normal(0, 30000, 100)

import pandas as pd

# 1. Creazione del DataFrame
# Utilizziamo i dati 'area' e 'prezzo_reale' generati in precedenza
df = pd.DataFrame({
    'Area_mq': area,
    'Prezzo_Euro': prezzo_reale
})

# 2. Arrotondamento per una visualizzazione più pulita
df = df.round(2)

# 3. Visualizzazione delle prime righe
print("Prime 5 righe del dataset:")
display(df.head(10))

# 4. Statistiche descrittive
print("\nAnalisi statistica del dataset:")
display(df.describe())
```

```python

# 2. Pre-calcolo della superficie di errore (per il grafico a destra)
# Creiamo una griglia di possibili valori per W e B per disegnare la "valle"
w_range = np.linspace(0, 4000, 50)
b_range = np.linspace(-100000, 200000, 50)
W_grid, B_grid = np.meshgrid(w_range, b_range)
MSE_grid = np.zeros(W_grid.shape)

for i in range(W_grid.shape[0]):
    for j in range(W_grid.shape[1]):
        preds = (area * W_grid[i,j]) + B_grid[i,j]
        MSE_grid[i,j] = np.mean((prezzo_reale - preds)**2)

# 3. Funzione di aggiornamento dinamico
def update_comparison(w, b):
    # Calcolo previsioni ed errore attuale
    y_pred = (area * w) + b
    current_mse = np.mean((prezzo_reale - y_pred)**2)
    
    clear_output(wait=True)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    
    # --- GRAFICO SINISTRA: Retta e Residui ---
    ax1.scatter(area, prezzo_reale, color='blue', marker='x', alpha=0.5, label='Dati reali')
    ax1.vlines(area, prezzo_reale, y_pred, color='gray', linestyle='--', alpha=0.2)
    x_line = np.array([40, 260])
    ax1.plot(x_line, (x_line * w) + b, color='red', lw=3, label=f'Modello attuale')
    ax1.set_title(f"Regressione Lineare\nMSE: {current_mse:,.0f}")
    ax1.set_ylim(0, 700000)
    ax1.set_xlabel("Area (mq)")
    ax1.set_ylabel("Prezzo (€)")
    ax1.legend()
    ax1.grid(True, alpha=0.2)
    
    # --- GRAFICO DESTRA: Loss Landscape (Mappa di calore dell'errore) ---
    # Usiamo una scala logaritmica per il colore per vedere meglio la valle
    cp = ax2.contourf(W_grid, B_grid, np.log10(MSE_grid), cmap='viridis', levels=20)
    # Disegniamo un punto rosso dove si trovano attualmente gli slider
    ax2.plot(w, b, marker='o', color='red', markersize=10, markeredgecolor='white', label='Tua posizione')
    ax2.set_title("Loss Landscape (Mappa dell'Errore)\nL'obiettivo è il centro della valle blu")
    ax2.set_xlabel("Parametro Peso (w)")
    ax2.set_ylabel("Parametro Bias (b)")
    ax2.legend()
    
    plt.tight_layout()
    plt.show()

# 4. Interazione
interact(update_comparison, 
         w=IntSlider(min=0, max=4000, step=50, value=1000, continuous_update=True),
         b=IntSlider(min=-100000, max=200000, step=5000, value=50000, continuous_update=True));
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SETUP DATI E FUNZIONI ---
np.random.seed(42)
# Generiamo dati dove la relazione vera è circa y = 2.5x
X_data = np.random.uniform(1, 5, 30)
w_true = 2.5
y_data = w_true * X_data + np.random.normal(0, 1, 30) # Aggiungiamo un po' di rumore

# Funzione per calcolare la MSE (la nostra parabola J(w))
def calculate_loss(w, X, y):
    predictions = w * X
    mse = np.mean((y - predictions)**2)
    return mse

# Funzione per calcolare il Gradiente (la derivata J'(w))
def calculate_gradient(w, X, y):
    N = len(y)
    predictions = w * X
    # La derivata di MSE rispetto a w per f(x)=wx è: -2/N * sum(x * error)
    gradient = -2/N * np.sum(X * (y - predictions))
    return gradient

# --- 2. ESECUZIONE DEL GRADIENT DESCENT ---
w_start = 0.0        # Partiamo da un peso sbagliato (lontano da 2.5)
learning_rate = 0.1 # Dimensione del passo
steps = 10           # Numero di passi da visualizzare

history_w = []
history_loss = []
w_current = w_start

# Ciclo che esegue l'algoritmo e salva la storia
for i in range(steps + 1):
    # 1. Calcola e salva la posizione attuale PRIMA di muoversi
    current_loss = calculate_loss(w_current, X_data, y_data)
    history_w.append(w_current)
    history_loss.append(current_loss)
    
    # 2. Calcola il gradiente e fai il passo (se non siamo all'ultimo step)
    if i < steps:
        grad = calculate_gradient(w_current, X_data, y_data)
        w_current = w_current - (learning_rate * grad)

# --- 3. CREAZIONE DELLA VISUALIZZAZIONE UNICA ---
plt.figure(figsize=(12, 8))

# A) Disegniamo la Parabola di sfondo (il "paesaggio")
w_grid = np.linspace(-0.5, 4.5, 200)
loss_grid = [calculate_loss(w, X_data, y_data) for w in w_grid]
plt.plot(w_grid, loss_grid, 'k--', linewidth=2, alpha=0.4, label='Funzione di Costo $J(w)$ (Parabola)')

# B) Disegniamo la traiettoria dell'algoritmo
# Linea che collega i punti
plt.plot(history_w, history_loss, color='red', linestyle='-', linewidth=2, alpha=0.6)
# I punti rossi (ogni fine step)
plt.scatter(history_w, history_loss, color='red', s=100, zorder=5, label='Step del Gradient Descent')

# C) Evidenziamo Inizio e Fine per chiarezza
plt.scatter(history_w[0], history_loss[0], color='green', s=150, marker='o', zorder=6, label='Start ($w_0$)')
plt.scatter(history_w[-1], history_loss[-1], color='blue', s=150, marker='*', zorder=6, label=f'End ($w_{{10}}$)')

# D) Aggiungiamo etichette ai primi passi per mostrare la direzione
for i in range(5):
    plt.text(history_w[i] + 0.05, history_loss[i] + 0.5, f'Step {i}', fontsize=9, color='darkred')

# Estetica
plt.title('Traiettoria del Gradient Descent sulla Funzione di Costo\n(Si nota come i passi si accorciano avvicinandosi al minimo)', fontsize=14)
plt.xlabel('Valore del Parametro Peso ($w$)', fontsize=12)
plt.ylabel('Errore (MSE)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.savefig('traiettoria_gradient_descent.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import time

# 1. Dataset (stessi dati per coerenza)
np.random.seed(42)
area = np.random.uniform(50, 250, 100)
prezzo_reale = (area * 2000) + 50000 + np.random.normal(0, 30000, 100)

# Normalizzazione veloce per stabilità numerica del gradiente
# (fondamentale per far funzionare bene l'algoritmo da zero)
x = area
y = prezzo_reale

# 2. Parametri Iniziali (Partiamo da valori "sbagliati")
w = 0.0
b = 0.0
learning_rate = 0.00001 # La velocità di discesa (alfa)
iterations = 10

# Storia per il grafico della Loss
history_w = []
history_b = []
history_loss = []

# 3. Ciclo di Gradient Descent con Animazione
for i in range(iterations):
    # --- LOGICA MATEMATICA ---
    y_pred = (x * w) + b
    error = y - y_pred
    mse = np.mean(error**2)
    
    # Calcolo dei gradienti (derivate parziali)
    # dw = d/dw di 1/n * sum(y - (wx+b))^2
    dw = -2 * np.mean(x * error)
    db = -2 * np.mean(error)
    
    # Aggiornamento dei pesi
    w = w - (learning_rate * dw)
    b = b - (learning_rate * db)
    
    # Salvataggio storia
    history_w.append(w)
    history_b.append(b)
    history_loss.append(mse)
    
    # --- VISUALIZZAZIONE DINAMICA ---
    if i % 2 == 0: # Aggiorniamo il grafico ogni 2 step per fluidità
        clear_output(wait=True)
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
        
        # Grafico 1: Retta che si adatta ai dati
        ax1.scatter(x, y, color='blue', marker='x', alpha=0.5)
        x_range = np.array([min(x), max(x)])
        ax1.plot(x_range, (x_range * w) + b, color='red', lw=3)
        ax1.set_title(f"Iterazione: {i}\n$f(x) = {w:.2f}x + {b:.2f}$")
        ax1.set_xlabel("Area")
        ax1.set_ylabel("Prezzo")
        ax1.grid(True, alpha=0.2)
        
        # Grafico 2: Discesa della Loss
        ax2.plot(history_loss, color='green', marker='o')
        ax2.set_title(f"Funzione di Costo (MSE)\nValore attuale: {mse:,.0f}")
        ax2.set_xlabel("Iterazione")
        ax2.set_ylabel("MSE")
        ax2.grid(True, alpha=1.0)
        
        plt.tight_layout()
        plt.show()
        time.sleep(2.0) # Pausa per l'animazione

print("Addestramento completato!")
```
