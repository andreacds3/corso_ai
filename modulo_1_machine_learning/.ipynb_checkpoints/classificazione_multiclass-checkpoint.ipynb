```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.datasets import make_blobs

# 1. Generazione di 3 classi di dati
X, y = make_blobs(n_samples=200, centers=3, n_features=2, random_state=42, cluster_std=1.5)

# 2. Addestramento Multinomiale (Softmax)
# multi_class='multinomial' forza l'uso della Softmax invece di tante sigmoidi
clf = LogisticRegression(multi_class='multinomial', solver='lbfgs')
clf.fit(X, y)

# 3. Visualizzazione
fig, ax = plt.subplots(figsize=(10, 8))

# Disegniamo i territori
DecisionBoundaryDisplay.from_estimator(
    clf, X,
    response_method="predict", # Mostra il verdetto finale (la regione)
    cmap=plt.cm.Paired,
    alpha=0.4,
    ax=ax
)

# Sovrapponiamo i punti originali con marker diversi
markers = ['o', 'X', '^']
colors = ['blue', 'red', 'green']
labels = ['Cerchi (0)', 'Croci (1)', 'Triangoli (2)']

for i in range(3):
    ax.scatter(X[y == i, 0], X[y == i, 1], 
               marker=markers[i], color=colors[i], 
               s=100, edgecolor='k', label=labels[i])

ax.set_title("Decision Boundaries Multiclasse (Softmax)", fontsize=15)
ax.legend()
plt.savefig('classificazione_multiclass.png')
plt.show()
```
