```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import matplotlib.cm as cm
import numpy as np

# --- 1. Generazione Dati Sintetici (4 Centri) ---
X_multi, y_multi = make_blobs(n_samples=300, centers=4, n_features=2, random_state=42, cluster_std=2.6)

# --- 2. Setup del Grafico ---
plt.figure(figsize=(10, 8))

# Definizione dei marker per le 4 classi
markers = ['o', 'X', '^', 's'] # Cerchio, Croce, Triangolo, Quadrato
labels = ['Classe 0 (O)', 'Classe 1 (X)', 'Classe 2 (Triangolo)', 'Classe 3 (Quadrato)']

# Usiamo una colormap per generare 4 colori distinti
colors = cm.viridis(np.linspace(0, 1, 4))

# --- 3. Loop di Plotting ---
for i in range(4):
    plt.scatter(X_multi[y_multi == i, 0], X_multi[y_multi == i, 1],
                marker=markers[i],       # Assegna il marker specifico
                color=colors[i],         # Assegna il colore specifico
                s=120,                   # Dimensione dei punti
                edgecolor='k',           # Bordino nero per farli risaltare (utile per cerchi e quadrati)
                linewidth=1.5,           # Spessore per le X e i triangoli
                label=labels[i])         # Etichetta per la legenda

# --- 4. Estetica ---
plt.title("Classificazione Multipla: 4 Categorie Distinte", fontsize=16, fontweight='bold')
plt.xlabel("Feature 1", fontsize=12)
plt.ylabel("Feature 2", fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Salvataggio opzionale
plt.savefig('multiclass_classification_shapes.png', dpi=300)

plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import DecisionBoundaryDisplay

# 1. Addestriamo un classificatore reale sui dati generati prima
# Usiamo il modello multinomiale (Softmax) per le 4 classi
clf = LogisticRegression(multi_class='multinomial', solver='lbfgs')
clf.fit(X_multi, y_multi)

# 2. Creazione del grafico con le aree di decisione
fig, ax = plt.subplots(figsize=(10, 8))

# Disegniamo le aree colorate (il "pensiero" del modello)
DecisionBoundaryDisplay.from_estimator(
    clf,
    X_multi,
    response_method="predict",
    cmap=plt.cm.Paired,
    alpha=0.3,
    ax=ax,
)

# 3. Sovrapponiamo i punti originali (i markers X, O, ^, s)
for i in range(4):
    ax.scatter(X_multi[y_multi == i, 0], X_multi[y_multi == i, 1],
                marker=markers[i], color=colors[i], s=100, 
                edgecolor='k', label=labels[i])

ax.set_title("Decision Boundaries: Il modello divide lo spazio in regioni", fontsize=16)
ax.legend()
plt.savefig('decision_boundaries.png', dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.datasets import make_blobs

# 1. Generazione dati binari (2 classi)
X, y = make_blobs(n_samples=100, centers=2, n_features=2, random_state=42, cluster_std=1.5)

# 2. Addestramento del modello (Regressione Logistica)
# Questo è il momento in cui il modello "impara" dove tracciare il confine
clf = LogisticRegression()
clf.fit(X, y)

# 3. Creazione del grafico
fig, ax = plt.subplots(figsize=(10, 8))

# Disegniamo le aree colorate (il confine di decisione)
# La zona blu rappresenta il territorio delle O, quella rossa delle X
DecisionBoundaryDisplay.from_estimator(
    clf,
    X,
    response_method="predict",
    cmap=plt.cm.RdBu,
    alpha=0.3,
    ax=ax,
)

# 4. Sovrapponiamo i punti originali
ax.scatter(X[y == 0, 0], X[y == 0, 1], marker='o', color='blue', s=100, edgecolor='k', label='Classe 0 (O)')
ax.scatter(X[y == 1, 0], X[y == 1, 1], marker='X', color='red', s=150, edgecolor='k', label='Classe 1 (X)')

# 5. Estetica
ax.set_title("Decision Boundary Binaria: Il confine tra 0 e 1", fontsize=16)
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.legend()
plt.grid(True, linestyle='--', alpha=0.3)
plt.savefig('binary_classification.png')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-10, 10, 200)
sigmoide = 1 / (1 + np.exp(-z))
step = np.where(z > 0, 1, 0)

plt.figure(figsize=(10, 6))
plt.plot(z, sigmoide, label='Sigmoide (Derivabile - Permette l\'apprendimento)', lw=3, color='green')
plt.plot(z, step, label='Step Function (Non derivabile - Blocca il gradiente)', lw=2, color='red', linestyle='--')

plt.axvline(0, color='black', alpha=0.3)
plt.axhline(0.5, color='gray', linestyle=':', alpha=0.5)
plt.title("Perché usiamo la Sigmoide?", fontsize=15)
plt.xlabel("Punteggio Lineare (z)")
plt.ylabel("Output")
plt.legend()
plt.grid(True, alpha=0.2)
plt.savefig('sigmoid.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.datasets import make_blobs

# 1. Generazione Dati
X, y = make_blobs(n_samples=150, centers=2, n_features=2, random_state=42, cluster_std=4.0)

# 2. Addestramento
clf = LogisticRegression()
clf.fit(X, y)

# 3. Grafico
fig, ax = plt.subplots(figsize=(10, 8))

# A) DISEGNIAMO IL GRADIENTE (lo sfondo sfumato)
# response_method="predict_proba" -> mostra la probabilità della Sigmoide
disp = DecisionBoundaryDisplay.from_estimator(
    clf, X,
    response_method="predict_proba",
    plot_method="contourf", # 'f' sta per filled (pieno)
    cmap="RdBu_r",
    alpha=0.7,
    ax=ax
)

# B) DISEGNIAMO IL CONFINE NETTO (la linea del 50%)
# Usiamo plot_method="contour" (solo linee) e fissiamo il livello a 0.5
DecisionBoundaryDisplay.from_estimator(
    clf, X,
    response_method="predict_proba",
    plot_method="contour",
    levels=[0.5], # Tracciamo solo dove la probabilità è esattamente 0.5
    colors="black",
    linestyles="--",
    linewidths=2,
    ax=ax
)

# C) SOVRAPPONIAMO I PUNTI
ax.scatter(X[y == 0, 0], X[y == 0, 1], marker='o', color='navy', s=100, edgecolor='k', label='Classe 0 (O)')
ax.scatter(X[y == 1, 0], X[y == 1, 1], marker='X', color='darkred', s=150, edgecolor='k', label='Classe 1 (X)')

# 4. ESTETICA
plt.colorbar(disp.surface_, ax=ax, label="Probabilità Classe 1 (X)")
ax.set_title("Decision Boundary con Probabilità Sigmoide", fontsize=15)
ax.legend()
plt.savefig('decision_boundary_prob.png')
plt.show()
```
