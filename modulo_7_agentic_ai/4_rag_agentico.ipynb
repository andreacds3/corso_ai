````python
from langgraph.graph import StateGraph, END
from typing import TypedDict, Annotated, Literal, List
import operator
from langchain_core.messages import AnyMessage, SystemMessage, HumanMessage, ToolMessage, AIMessage
from langgraph.graph.message import add_messages
from pydantic import BaseModel, Field
import json

# 1. Output Strutturato
class LivelloResult(BaseModel):
    livello_min_consigliato: int 
    livello_max_consigliato: int
    motivazione: str = Field(description="Breve spiegazione del perché questi livelli sono scelti")
    fonte_livello: str = Field(description="capitoli_recuperati, lezioni_recuperate, stima_autonoma, o sintesi_ibrida")

# 2. Stato dell'Agente (Memorie Isolate)
class AgentState(TypedDict):
    lesson: str
    # 'add_messages' gestisce automaticamente l'append e la corretta sequenza AI -> Tool
    messages_capitoli: Annotated[list, add_messages]
    messages_lezioni: Annotated[list, add_messages]
    answer_capitoli: str
    answer_lezioni: str
    final_result: AIMessage


class AgenticWorkflow:

    def __init__(self, model, tools_capitoli, tools_lezioni, system_message_capitoli="", system_message_lezioni="", system_message_revisore=""):
        self.system_message_capitoli = system_message_capitoli
        self.system_message_lezioni = system_message_lezioni
        self.system_message_revisore = system_message_revisore
        
        # Mappe dei tool
        self.tools_capitoli_dict = {t.name: t for t in tools_capitoli} #RAG CAPITOLI
        self.tools_lezioni_dict = {t.name: t for t in tools_lezioni} #RAG LEZIONI
        
        # Binding
        self.model_with_rag_capitoli = model.bind_tools(tools_capitoli)
        self.model_with_rag_lezioni = model.bind_tools(tools_lezioni)
        self.model_reflection = model.with_structured_output(LivelloResult) 

        graph = StateGraph(AgentState) 
        
        graph.add_node("llm_capitoli", self.call_llm_rag_capitoli)
        graph.add_node("action_capitoli", self.take_action_rag_capitoli)
        
        graph.add_node("llm_lezioni", self.call_llm_rag_lezioni)
        graph.add_node("action_lezioni", self.take_action_rag_lezioni)
        
        graph.add_node("reflection", self.call_reflection)

        # FLUSSO 1: Capitoli
        graph.set_entry_point("llm_capitoli")
        graph.add_conditional_edges(
            "llm_capitoli",
            self.should_continue_capitoli,
            {
                "continue": "action_capitoli",
                "end": "llm_lezioni"
            }
        )
        graph.add_edge("action_capitoli", "llm_capitoli")

        # FLUSSO 2: Lezioni
        graph.add_conditional_edges(
            "llm_lezioni",
            self.should_continue_lezioni,
            {
                "continue": "action_lezioni",
                "end": "reflection"
            }
        )
        graph.add_edge("action_lezioni", "llm_lezioni")
        
        # FLUSSO 3: Revisore
        graph.add_edge("reflection", END)

        self.graph = graph.compile()


    # --- METODI LLM ---

    def call_llm_rag_capitoli(self, state: AgentState):
        print('--- LLM RAG CAPITOLI ---')
        messages = state.get('messages_capitoli', [])
        
        # Inizializza la conversazione solo se è il primo ciclo
        if not messages:
            if self.system_message_capitoli:
                messages.append(SystemMessage(content=self.system_message_capitoli))
            messages.append(HumanMessage(content=f"Analizza il livello di questa lezione:\n\n{state['lesson']}"))

        message = self.model_with_rag_capitoli.invoke(messages)
        
        # Ritorna per appendere il messaggio allo storico
        if not message.tool_calls:
            # Se ha finito di usare i tool, salviamo la stringa finale in answer_capitoli
            #print(message.content[0]['text'])
            print(message.content[0]['text'])

            return {"messages_capitoli": [message], "answer_capitoli": message.content}
        return {"messages_capitoli": [message]}

    def call_llm_rag_lezioni(self, state: AgentState):
        print('--- LLM RAG LEZIONI ---')
        messages = state.get('messages_lezioni', [])
        
        if not messages:
            if self.system_message_lezioni:
                messages.append(SystemMessage(content=self.system_message_lezioni))
            messages.append(HumanMessage(content=f"Analizza il livello di questa lezione:\n\n{state['lesson']}"))

        message = self.model_with_rag_lezioni.invoke(messages)
        
        if not message.tool_calls:
            #print(message.content[0]['text'])
            print(message.content[0]['text'])
            return {"messages_lezioni": [message], "answer_lezioni": message.content}
        return {"messages_lezioni": [message]}

    def call_reflection(self, state: AgentState):
        print('--- LLM REFLECTION (REVISORE) ---')
        
        prompt_arbitraggio = (
            f"Devi arbitrare il livello scolare per questa descrizione:\n{state['lesson']}\n\n"
            f"=== ANALISI DEI LIBRI DI TESTO (CAPITOLI) ===\n{state['answer_capitoli']}\n\n"
            f"=== ANALISI DELL'APPLICAZIONE PRATICA (LEZIONI) ===\n{state['answer_lezioni']}\n\n"
            "Applica le regole di arbitraggio ed emetti il risultato JSON finale."
        )
        
        messages = []
        if self.system_message_revisore:
             messages.append(SystemMessage(content=self.system_message_revisore))
        messages.append(HumanMessage(content=prompt_arbitraggio))
        
        result = self.model_reflection.invoke(messages)
        
        print("RISULTATO FINALE:", result.dict())
        return {"final_result": result}


    # --- METODI CONDIZIONALI (ROUTING) ---

    def should_continue_capitoli(self, state: AgentState):
        last_message = state['messages_capitoli'][-1]
        if last_message.tool_calls:
            return "continue"
        return "end"

    def should_continue_lezioni(self, state: AgentState):
        last_message = state['messages_lezioni'][-1]
        if last_message.tool_calls:
            return "continue"
        return "end"


    # --- METODI ACTION (TOOL) ---

    def take_action_rag_capitoli(self, state: AgentState):
        tool_calls = state['messages_capitoli'][-1].tool_calls
        results = []
        for t in tool_calls:
            if not t['name'] in self.tools_capitoli_dict:
                result = "bad tool name, retry"
            else:
                result = self.tools_capitoli_dict[t['name']].invoke(t['args'])
            results.append(ToolMessage(tool_call_id=t['id'], name=t['name'], content=str(result)))
        return {'messages_capitoli': results}
    
    def take_action_rag_lezioni(self, state: AgentState):
        tool_calls = state['messages_lezioni'][-1].tool_calls
        results = []
        for t in tool_calls:
            if not t['name'] in self.tools_lezioni_dict:
                result = "bad tool name, retry"
            else:
                result = self.tools_lezioni_dict[t['name']].invoke(t['args'])
            results.append(ToolMessage(tool_call_id=t['id'], name=t['name'], content=str(result)))
        return {'messages_lezioni': results}
````

````python
SYSTEM_MESSAGE_LEZIONI = """
# SYSTEM MESSAGE: ANALISTA PEDAGOGICO (VALUTAZIONE LEZIONI)

Sei un analista esperto di architetture didattiche. Il tuo compito è determinare il target scolare (scala 0-12) di una lezione, confrontando la sua descrizione testuale con i metadati di lezioni simili recuperate tramite RAG.

A differenza dei rigidi testi scolastici, le lezioni possono variare per tono e approccio. Il tuo compito è valutare il tono della lezione per decidere come applicare i dati del RAG, ma **non puoi mai inventare livelli fuori dal range fornito dal RAG**.

## 1. CRITERI DI CALCOLO (ANCORATI AL RAG)

* **LIVELLO MINIMO (Soglia di Accessibilità):**
    * **Identifica il Tono:** La lezione è divulgativa/pop (es. YouTube, metafore) o tecnica/accademica (es. analisi del testo, lezioni universitarie)?
    * **Applica il Vincolo RAG:**
        * Se è **tecnica**, applica il *Max dei Minimi* tra le lezioni RAG pertinenti (cerca il rigore).
        * Se è **divulgativa**, cerca tra le lezioni RAG pertinenti quella con il "Livello Minimo" più basso e usa quello.
        * **DIVIETO ASSOLUTO:** Non puoi **mai** scendere sotto il "Livello Minimo" più basso presente tra le lezioni RAG pertinenti. Se il RAG più basso per Freud è 12, il tuo minimo DEVE essere 12, anche se il video è su YouTube.

* **LIVELLO MASSIMO (Tetto Curricolare - Rigido):**
    * **REGOLA DEL TETTO ASSOLUTO:** Il livello Massimo della lezione **NON DEVE MAI superare** il valore massimo indicato dalle lezioni RAG pertinenti (*Max dei Massimi*).
    * **Divieto di Estensione:** Se le lezioni RAG pertinenti si fermano a 10, il tuo massimo **deve essere 10**. Il RAG detta i limiti del curriculum.

## 2. FILTRO DI PERTINENZA (CONTESTUALE)
* **Match di Sostanza:** Usa solo le lezioni RAG che affrontano lo stesso argomento principale, periodo storico o autore centrale. Non confondere il *mezzo* (es. YouTube) con il *messaggio* (es. psicoanalisi clinica).
* **Contesto del Formato:** Se la descrizione indica un riassunto o una sintesi, ignora le estensioni massime di lezioni RAG che sono palesemente monografie di livello universitario.

## 3. GESTIONE FALLIMENTO (STIMA AUTONOMA)
* Usa la "stima_autonoma" **SOLO** se non esiste nemmeno una lezione RAG pertinente all'argomento centrale. Se i dati RAG ci sono, devi usarli come recinto invalicabile.
* *Reference per stima autonoma (solo come ultima ratio):* 0-4 (Primaria), 5-7 (Secondaria I grado), 8-12 (Secondaria II grado).

## 4. SCHEMA DI OUTPUT (PER AGENTE REVISORE)
Rispondi rigorosamente seguendo questa struttura testuale:

* **ANALISI DEL TONO:** (Definisci se è divulgativo o tecnico, prestando attenzione al contenuto reale)
* **SELEZIONE RAG:** (Indica i titoli/ID usati e giustifica gli scarti)
* **LIVELLO MINIMO:** (Solo numero. Deve essere giustificato dai dati RAG)
* **LIVELLO MASSIMO:** (Solo numero. Deve essere giustificato dai dati RAG)
* **FONTE:** (lezioni_recuperate / stima_autonoma)
* **MOTIVAZIONE:** (Spiega come hai usato il tono per scegliere tra i minimi del RAG e conferma l'applicazione della Regola del Tetto)
"""
````

````python
SYSTEM_MESSAGE_REVISORE = """"
# SYSTEM MESSAGE: REVISORE PEDAGOGICO (ARBITRO FINALE)

Sei il Revisore Pedagogico Finale. Il tuo compito è stabilire il target scolare definitivo (0-12) risolvendo i conflitti tra due analisi pre-compilate:
1. **Analisi Capitoli (Libri di Testo):** Rigida, ancorata al curriculum ufficiale.
2. **Analisi Lezioni (Applicazione Pratica):** Flessibile, sensibile al formato divulgativo o all'accessibilità.

## 1. REGOLE DI ARBITRAGGIO (I 3 SCENARI)
Devi classificare la descrizione in uno di questi tre scenari e applicare la relativa regola per formare il range (Minimo-Massimo).

* **SCENARIO A: Autore o Testo Specifico (Match Esatto)**
    * *Cosa include:* Figure storiche, opere precise, dottrine tecniche (es. "Transfert in Freud", "Femminismo di Mary Wollstonecraft", "Sillogismo").
    * *Chi vince:* **Vince il match più preciso in blocco.** Identifica quale agente ha trovato il nome/concetto esatto nel RAG. Adotta il suo Minimo e il suo Massimo senza fare mix. Se entrambi hanno il match esatto, l'Analisi Capitoli ha la priorità.

* **SCENARIO B: Temi Universali ed Esistenziali (Sintesi Espansa)**
    * *Cosa include:* Concetti umani generali, estetica, etica, astrazioni non relegate a un singolo autore (es. "La bellezza", "Il simbolo", "L'amore", "La felicità").
    * *Chi vince:* **Nessuno vince in blocco, si fa una sintesi ibrida.** Questi temi hanno una bassa barriera d'ingresso ma un altissimo potenziale. Adotta il **Minimo più basso** proposto tra le due analisi (spesso 8, dall'Analisi Lezioni) e il **Massimo più alto** (spesso 12, dall'Analisi Capitoli).

* **SCENARIO C: Pop-Culture e Attualità (Priorità Lezioni)**
    * *Cosa include:* Analisi di film, manga, serie TV, dibattiti contemporanei non curricolari (es. "Filosofia di Matrix", "Joker", "IA").
    * *Chi vince:* **Vince l'Analisi Lezioni in blocco.** I libri di testo (Capitoli) sono fuori contesto. Adotta sia il Minimo che il Massimo dell'Analisi Lezioni.

## 2. FORMATO DI OUTPUT (JSON)
Devi produrre ESCLUSIVAMENTE un oggetto JSON valido. Non aggiungere testo fuori dal JSON.

```json
{
  "livello_min_consigliato": int,
  "livello_max_consigliato": int,
  "motivazione": "1. Dichiara lo Scenario (A, B o C). 2. Giustifica brevemente l'applicazione della regola corrispondente sui dati forniti.",
  "fonte_livello": "capitoli_recuperati" O "lezioni_recuperate" O "sintesi_ibrida"
}
"""
````

````python
@tool
def get_similar_chapter(descrizione_lezione: str):
    """
    Usa un RAG per recuperare capitoli di libri che trattano argomenti afferenzi la descrizione_lezione.
    Ritorna un testo con il titolo del capitolo, la descrizione e il livello minimo e massimo del capitolo. I livelli sono distribuiti nel seguente modo:
    0-4 dalla prima elementare alla quinta elementare. 
    5-7 dalla prima media alla terza media.
    8-12 dal primo superiore al quinto superiore.
    """
    dense_vector = next(dense_embedding_model.embed(descrizione_lezione)) #next(dense_embedding_model.query_embed(query))
    sparse_vector = next(bm25_embedding_model.embed(descrizione_lezione))#next(bm25_embedding_model.query_embed(query))
    late_interaction_vector = next(late_interaction_embedding_model.embed(descrizione_lezione)) #next(late_interaction_embedding_model.query_embed(query))
    prefetch = [
            models.Prefetch(
                query=dense_vector,
                using="dense",
                limit=20,
            ),
            models.Prefetch(
                query=models.SparseVector(**sparse_vector.as_object()),
                using="bm25",
                limit=20,
            ),
        ]
    results = client.query_points(
             "rag_capitoli",
            prefetch=prefetch,
            query=late_interaction_vector,
            using="colbert",
            with_payload=True,
            limit=20,
    )
    
    points = results.points
    results = model.rerank(descrizione_lezione, [point.payload['descr'] for point in points])
    results = [point.payload for point in points if point.payload['descr'] in [result['document'] for result in results[:10]]]
    return '\n\n'.join([f'Descrizione: {res["descr"]}\nTitolo: {res["titolo"]}\nLivello Minimo: {res["liv_min"]}\nLivello Massimo: {res["liv_max"]}' for res in results])
````

````python
@tool
def get_similar_lesson(descrizione_lezione: str):
    """
    Usa un RAG per recuperare lezioni che trattano argomenti afferenzi la descrizione_lezione.
    Ritorna un testo con il titolo della lezione, la descrizione e il livello minimo e massimo della lezione. I livelli sono distribuiti nel seguente modo:
    0-4 dalla prima elementare alla quinta elementare. 
    5-7 dalla prima media alla terza media.
    8-12 dal primo superiore al quinto superiore.
    """
    dense_vector = next(dense_embedding_model.embed(descrizione_lezione)) #next(dense_embedding_model.query_embed(query))
    sparse_vector = next(bm25_embedding_model.embed(descrizione_lezione))#next(bm25_embedding_model.query_embed(query))
    late_interaction_vector = next(late_interaction_embedding_model.embed(descrizione_lezione)) #next(late_interaction_embedding_model.query_embed(query))
    prefetch = [
            models.Prefetch(
                query=dense_vector,
                using="dense",
                limit=20,
            ),
            models.Prefetch(
                query=models.SparseVector(**sparse_vector.as_object()),
                using="bm25",
                limit=20,
            ),
        ]
    results = client.query_points(
             "rag_lezioni",
            prefetch=prefetch,
            query=late_interaction_vector,
            using="colbert",
            with_payload=True,
            limit=20,
    )
    
    points = results.points
    results = model.rerank(descrizione_lezione, [point.payload['descr'] for point in points])
    results = [point.payload for point in points if point.payload['descr'] in [result['document'] for result in results[:10]]]
    return '\n\n'.join([f'Descrizione: {res["descr"]}\nTitolo: {res["titolo"]}\nLivello Minimo: {res["liv_min"]}\nLivello Massimo: {res["liv_max"]}' for res in results])
````

````python
workflow = AgenticWorkflow(model=gemini_model, system_message_capitoli=SYSTEM_MESSAGE_CAPITOLI, 
                           system_message_lezioni = SYSTEM_MESSAGE_LEZIONI, 
                           system_message_revisore=SYSTEM_MESSAGE_REVISORE,
                          tools_capitoli=[get_similar_chapter], tools_lezioni=[get_similar_lesson])
````
